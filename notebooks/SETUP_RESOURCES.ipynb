```
# ============================================================
# 🧠 Télécharger Depth Anything V2 (~1.3GB)
# ============================================================
import os
model_path = f"{AI_MODELS}/depth_anything_v2/depth_anything_v2_vitl.pth"
model_url = "https://huggingface.co/depth-anything/Depth-Anything-V2-Large/resolve/main/depth_anything_v2_vitl.pth"

if os.path.exists(model_path):
    print(f"✅ Depth Anything V2 déjà présent ({os.path.getsize(model_path)/1e9:.2f} GB)")
else:
    print("⏳ Téléchargement Depth Anything V2 (~1.3GB)...")
    !wget -q --show-progress -O "{model_path}" "{model_url}"
    print(f"✅ Téléchargé: {os.path.getsize(model_path)/1e9:.2f} GB")
```

```
# ============================================================
# 🔮 Télécharger Real-ESRGAN (~64MB)
# ============================================================
model_path = f"{AI_MODELS}/esrgan/RealESRGAN_x4plus.pth"
model_url = "https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth"

if os.path.exists(model_path):
    print(f"✅ Real-ESRGAN déjà présent ({os.path.getsize(model_path)/1e6:.1f} MB)")
else:
    print("⏳ Téléchargement Real-ESRGAN (~64MB)...")
    !wget -q --show-progress -O "{model_path}" "{model_url}"
    print(f"✅ Téléchargé: {os.path.getsize(model_path)/1e6:.1f} MB")
```

```
# ============================================================
# 📊 Rapport Final
# ============================================================
print("="*60)
print("📊 RAPPORT SETUP RESSOURCES")
print("="*60)

models = [
    ("Depth Anything V2", f"{AI_MODELS}/depth_anything_v2/depth_anything_v2_vitl.pth"),
    ("YOLOv8x", f"{AI_MODELS}/yolov8/yolov8x.pt"),
    ("SAM ViT-H", f"{AI_MODELS}/sam/sam_vit_h_4b8939.pth"),
    ("Real-ESRGAN", f"{AI_MODELS}/esrgan/RealESRGAN_x4plus.pth"),
]

total_size = 0
all_ok = True
for name, path in models:
    if os.path.exists(path):
        size = os.path.getsize(path)
        total_size += size
        print(f"✅ {name}: {size/1e9:.2f} GB")
    else:
        print(f"❌ {name}: MANQUANT")
        all_ok = False

print("="*60)
print(f"💾 Total: {total_size/1e9:.2f} GB")
print("="*60)

if all_ok:
    print("\n🚀 SETUP COMPLET - Prêt pour production!")
    print("\n📖 Prochaine étape: GUIDE_PRODUCTION.md")
else:
    print("\n⚠️ Certains modèles manquent - Réexécuter les cellules")
```
