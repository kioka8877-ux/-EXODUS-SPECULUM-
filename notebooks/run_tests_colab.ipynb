## 4. Run All Tests with Report

```python
# Exécuter tous les tests avec rapport JSON
!python -m pytest tests/ -v --tb=short --json-report --json-report-file=test_report.json

import json
try:
    with open("test_report.json") as f:
        report = json.load(f)
    
    print("\n" + "="*60)
    print("📊 RÉSULTATS DES TESTS")
    print("="*60)
    print(f"   Total:   {report['summary']['total']}")
    print(f"   Passed:  {report['summary'].get('passed', 0)} ✅")
    print(f"   Failed:  {report['summary'].get('failed', 0)} ❌")
    print(f"   Skipped: {report['summary'].get('skipped', 0)} ⏭️")
    print(f"   Duration: {report['summary'].get('duration', 0):.2f}s")
    print("="*60)
    
    # Afficher les tests échoués
    failed_tests = [t for t in report.get('tests', []) if t.get('outcome') == 'failed']
    if failed_tests:
        print("\n❌ Tests échoués:")
        for test in failed_tests:
            print(f"   - {test['nodeid']}")
            if 'call' in test and 'longrepr' in test['call']:
                print(f"     {test['call']['longrepr'][:100]}...")
except FileNotFoundError:
    print("⚠️ Rapport non généré (pytest-json-report peut ne pas être installé)")
```

## 7. Test Coverage Summary

```python
# Résumé de couverture des tests
from pathlib import Path

unit_tests = list(Path("tests/unit").glob("test_*.py"))
integration_tests = list(Path("tests/integration").glob("test_*.py"))

print("📋 Test Coverage Summary")
print("=" * 50)
print(f"\nUnit Tests: {len(unit_tests)} files")
for t in sorted(unit_tests):
    print(f"   - {t.name}")

print(f"\nIntegration Tests: {len(integration_tests)} files")
for t in sorted(integration_tests):
    print(f"   - {t.name}")

# Count test functions
total_tests = 0
for test_file in unit_tests + integration_tests:
    content = test_file.read_text()
    test_count = content.count("def test_")
    total_tests += test_count

print(f"\n📊 Total test functions: ~{total_tests}")
print("=" * 50)
```
