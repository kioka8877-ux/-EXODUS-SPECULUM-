```
import os

ROOT = '/content/drive/MyDrive/EXODUS-SPECULUM'

frigates = [
    ('FRIGATE_00_CORTEX', 'Intelligence IA - Gemini 1.5 Pro'),
    ('FRIGATE_01_SCANNER', 'Extraction + Depth Estimation'),
    ('FRIGATE_02_SCENOGRAPHE', 'Génération géométrie 3D'),
    ('FRIGATE_03_PROJECTIONNISTE', 'Camera Projection Mapping'),
    ('FRIGATE_04_LOGISTIQUE', 'Asset replacement'),
    ('FRIGATE_05_DIRECTEUR_PHOTO', 'Camera animation'),
    ('FRIGATE_06_ALCHIMISTE', 'Rendu + Upscaling'),
    ('FRIGATE_07_PORTE_AVIONS', 'Assemblage final'),
]

print('📂 Création EXODUS-SPECULUM...')
print('='*50)

for name, desc in frigates:
    for sub in ['CODEBASE', 'INPUT', 'OUTPUT']:
        os.makedirs(f'{ROOT}/{name}/{sub}', exist_ok=True)
    print(f'✅ {name}/')

# SHARED_RESOURCES
for model in ['depth_anything_v2', 'yolov8', 'sam', 'esrgan', 'rife']:
    os.makedirs(f'{ROOT}/SHARED_RESOURCES/AI_MODELS/{model}', exist_ok=True)
print('✅ SHARED_RESOURCES/AI_MODELS/')

os.makedirs(f'{ROOT}/ASSETSHUB', exist_ok=True)
print('✅ ASSETSHUB/')

print('='*50)
print(f'🚀 Structure créée: {ROOT}')
```
