Crée toute l'arborescence de dossiers sur Google Drive.

| 📁 Dossiers | 7 unités + AI_MODELS |

```
from google.colab import drive
drive.mount('/content/drive')
print('✅ Drive monté!')
```

```
import os

ROOT = '/content/drive/MyDrive/DRIVE_EXODUS_V2'

structure = {
    '00_CORTEX_HQ': ['CODEBASE', 'IN_VIDEO_SOURCE', 'OUT_PRODUCTION_PLAN'],
    '01_ANIMATION_ENGINE': ['CODEBASE', 'IN_CORTEX_JSON', 'IN_MIXAMO_BASE', 'OUT_MOTION_DATA'],
    '02_LOGISTICS_DEPOT': ['CODEBASE', 'IN_MOTION_DATA', 'IN_ROBLOX_AVATAR', 'IN_PROPS_LIBRARY', 'OUT_BAKED_ACTORS'],
    '03_SCENOGRAPHY_DOCK': ['CODEBASE', 'IN_MAP_RAW/hdri_library', 'IN_MAP_RAW/environment_assets', 'IN_CORTEX_JSON', 'OUT_PREMIUM_SCENE'],
    '04_PHOTOGRAPHY_WING': ['CODEBASE', 'IN_VIDEO_SOURCE', 'IN_SCENE_REF', 'OUT_CAMERA_LOGIC'],
    '05_ALCHEMIST_LAB': ['CODEBASE', 'IN_RAW_FRAMES', 'LUTS', 'OUT_FINAL_FRAMES'],
    '06_AIRCRAFT_CARRIER': ['CODEBASE', 'IN_ASSEMBLY_KIT', 'OUT_FINAL_MOVIE'],
    'EXODUS_AI_MODELS': ['BLENDER', 'EMOCA', 'RIFE', 'REALESRGAN', 'McPrep', 'HDRi'],
}

print('📂 Création structure DRIVE_EXODUS_V2...')
print('='*50)

for unit, subdirs in structure.items():
    for subdir in subdirs:
        path = f'{ROOT}/{unit}/{subdir}'
        os.makedirs(path, exist_ok=True)
    print(f'✅ {unit}/ ({len(subdirs)} sous-dossiers)')

print('='*50)
print(f'✅ Structure créée: {ROOT}')
```
